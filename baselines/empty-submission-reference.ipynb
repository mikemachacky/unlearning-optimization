# Empty Submission Reference

This notebook generates a sample submission with an empty submission file.

```python
import os
import pandas as pd
from tqdm import tqdm
from pathlib import Path

os.makedirs("empty_submission", exist_ok=True)
SUBMISSION_PATH = "empty_submission/submission.csv"

TEST_DIR = "../resources/test_set"
test_files = sorted(Path(TEST_DIR).glob("*.png"))

rows = []
for img_path in tqdm(test_files):
    # Use a single space for empty predictions so Kaggle's null-check passes.
    rows.append({"image_id": img_path.stem, "prediction_string": " "})

submission = pd.DataFrame(rows)
submission.insert(0, "id", range(len(submission)))
submission.to_csv(SUBMISSION_PATH, index=False)
print(f"Wrote {SUBMISSION_PATH}  ({len(submission)} rows)")
submission.head()
```
